```python
from fdip import FDIP
```

```python
p = FDIP("slagheap.res")
print(p)
```

```python
p.autoFilter()
print(p)
```

```python
p.filter(kmax=30000)
p.filter(m=23)
p.showSingleFrequencyData(0.3);
p.filter(fmax=50.)
```

```python
p.basename += '_corr'
if 0: # generate two pdf files
    p.generateDataPDF()
    p.generateSpectraPDF()
```

```python
p.saveData()
```

```python
p.singleInversion(5.0, zWeight=0.3, verbose=True)
```

```python
p.showSingleResult(imax=200);
```
